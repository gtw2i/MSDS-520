# Matrices: Arithmetic and Special Types

**This notebook teaches matrices from scratch, no prior exposure is assumed.**

A **matrix** is a rectangular grid of numbers. It can hold a table of data (patients × features), the pixels of an image, or the coefficients of a linear system. Before matrices can *do* anything, we need the basics: how to read an entry, add and scale matrices, flip one with the transpose, and recognize a few special matrices that appear everywhere. This notebook covers exactly that, **multiplication comes next**.

## Learning objectives

- Read a matrix's **entries** $a_{ij}$ and its **dimensions** $m \times n$.
- **Add** matrices and multiply by a **scalar**.
- Take a **transpose**, and recognize the **zero, identity, diagonal, and symmetric** matrices.

## Background

A matrix $A = [a_{ij}]$ has $m$ rows and $n$ columns; the entry $a_{ij}$ sits in row $i$, column $j$ (so $a_{12}$ is row 1, column 2). Two matrices of the same shape **add** entry by entry, and a **scalar** multiplies every entry. The **transpose** $A^\top$ swaps rows and columns. Special matrices worth knowing by name:

- **zero** (all entries 0), **identity** $I$ (1's on the diagonal, else 0),
- **diagonal** (nonzero only on the diagonal), **symmetric** ($A = A^\top$, a mirror across the diagonal).

## 1. Entries and dimensions

The matrix below has 2 rows and 2 columns, so its dimension is $2 \times 2$. The entry $a_{12}$ (row 1, column 2) is $2$. (Python counts from 0, so that entry is `A[0, 1]`.)

```python
A = np.array([[1, 2],
              [3, 4]])

helpers.show(A, name="A")
print("dimensions (rows, cols):", A.shape)
print("entry a_12 (row 1, col 2):", A[0, 1])
```

## 2. Addition and scalar multiplication

Same-shape matrices add entry by entry; a scalar multiplies every entry.

$$\begin{bmatrix}1&2\\3&4\end{bmatrix} + \begin{bmatrix}5&6\\7&8\end{bmatrix} = \begin{bmatrix}6&8\\10&12\end{bmatrix}, \qquad 2\begin{bmatrix}1&2\\3&4\end{bmatrix} = \begin{bmatrix}2&4\\6&8\end{bmatrix}.$$

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

helpers.show(A + B, name="A + B")
helpers.show(2 * A, name="2A")
```

## 3. The transpose

The **transpose** flips a matrix across its diagonal, turning rows into columns:

$$\begin{bmatrix}1&2\\3&4\end{bmatrix}^\top = \begin{bmatrix}1&3\\2&4\end{bmatrix}.$$

```python
A = np.array([[1, 2], [3, 4]])
helpers.show(A.T, name="A^T")
```

## 4. Special matrices

A few matrices come up constantly, and it is worth being able to name them on sight:

- **zero**: every entry 0;
- **identity** $I$: 1's on the diagonal, 0 everywhere else;
- **diagonal**: nonzero only on the diagonal;
- **symmetric**: equal to its own transpose, $A = A^\top$ (a mirror image across the diagonal).

For now these are just shapes to recognize. Each earns its name once **matrix multiplication** arrives in the next notebook, that is where the identity turns out to be the "do nothing" matrix and a diagonal matrix turns out to scale each axis.

```python
zero = np.zeros((2, 2))
identity = np.eye(2)
diagonal = np.array([[2, 0], [0, 5]])
symmetric = np.array([[1, 4], [4, 3]])

helpers.show(zero, name="zero")
helpers.show(identity, name="I")
helpers.show(diagonal, name="diagonal")
helpers.show(symmetric, name="symmetric")
print("is symmetric equal to its transpose?", np.array_equal(symmetric, symmetric.T))
```

## 5. Application: images are matrices

A grayscale image is just a matrix of brightness values. Here a small "plus" shape is a $3\times3$ matrix. **Adding** another matrix to it changes the brightnesses, a simple way to boost contrast (make the bright parts brighter and the dark parts darker).

```python
plus = np.array([[1, 2, 1],
                 [2, 2, 2],
                 [1, 2, 1]])

boost = np.array([[-1, 1, -1],
                  [ 1, 1,  1],
                  [-1, 1, -1]])

high_contrast = plus + boost
helpers.show(high_contrast, name="high-contrast image")

# --- graphics: show both as images (mechanics only) ---
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(plus, cmap="gray"); axes[0].set_title("original")
axes[1].imshow(high_contrast, cmap="gray"); axes[1].set_title("higher contrast")
plt.show()
```
