```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("dark_background")
```

```python
n_points = 100

x = np.random.uniform(-1,1,n_points)
y = np.random.uniform(-1,1,n_points)

for i in range(n_points):
    if x[i]**2 + y[i]**2 < 1:
        plt.plot( x[i], y[i], 'r.' )
    else:
        plt.plot( x[i], y[i], 'b.' )
    # end
# end

plt.gca().set_aspect("equal")
```

```python
def CalcPi(n_points):
    x = np.random.uniform(-1,1,n_points)
    y = np.random.uniform(-1,1,n_points)
    
    n_points_in_circle = 0.0
    for i in range(n_points):
        if x[i]**2 + y[i]**2 < 1:
            n_points_in_circle+=1
        # end
    # end
    
    return 4 * n_points_in_circle / n_points
# end
```

```python
CalcPi(10000)
```

```python
n_points_array = 2**np.arange(4,16,2)
print(n_points_array)
print()

n_repeat = 100

pis = []
for i in range(n_repeat):
    pis.append([])
    for j in n_points_array:
        pis[i].append( CalcPi(j) )
    # ene
# end
pis = np.array(pis)

df = pd.DataFrame(pis, columns=n_points_array)
df
```

```python
fig = plt.figure(figsize=(4,4))
df.boxplot()

plt.plot(range(1,len(n_points_array)+1),
         np.ones(len(n_points_array))*np.pi, 'r' )
plt.grid(False)
```
