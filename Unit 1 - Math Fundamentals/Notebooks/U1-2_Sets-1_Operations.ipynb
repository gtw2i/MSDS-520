# Sets and Their Operations

**This notebook teaches sets from scratch, no prior exposure is assumed.**

A **set** is one of the most basic ideas in all of mathematics: it is simply a **collection of distinct things**. The things can be numbers (the set of even numbers), objects, or, as we will use throughout, people (the set of patients who have a particular condition). Once a group is described as a set, we can ask precise questions about it: who belongs to two groups at once? who belongs to neither? how many people are in the combined group? The tools for answering those questions are the **set operations**, and they are the subject of this notebook.

## Learning objectives

- Describe a group as a **set** and test whether something is a **member** of it.
- Use the ideas of **subset** ($\subseteq$), the **empty set** ($\varnothing$), and the **universe** ($U$).
- Combine sets with the four operations: **union, intersection, complement, and difference**.
- Count the elements of a set, its **cardinality**, $|A|$.

## Background

A **set** is a collection of distinct items called its **elements** (or members). If an item $x$ belongs to a set $A$ we write $x \in A$ ("$x$ is in $A$"); if it does not, $x \notin A$. Order and repetition do not matter, $\{1, 2\}$ and $\{2, 1, 1\}$ describe the same set.

We usually work inside a fixed **universe** $U$: the set of everything currently under discussion (here, all the patients in our study). For sets $A$ and $B$ inside $U$:

$$A \cup B \ (\text{union: in } A \textbf{ or } B), \quad A \cap B \ (\text{intersection: in } A \textbf{ and } B), \quad A^c = U \setminus A \ (\textbf{not}\text{ in } A), \quad A \setminus B \ (\text{in } A \textbf{ but not } B).$$

Two more basic ideas: $A$ is a **subset** of $B$ (written $A \subseteq B$) when every element of $A$ is also in $B$; the **empty set** $\varnothing$ is the set with no elements at all. The number of elements in $A$ is its **cardinality**, written $|A|$.

## 1. A small group of patients

Everything below is built on one small example: eight patients, labelled P1 to P8. For each patient we record whether they have three conditions, diabetes, hypertension, and obesity, as `True` (yes) or `False` (no). This is just a table of data: each **row** is one patient. The whole group of eight is our **universe** $U$, the collection of everyone we are talking about.

```python
# Build a tiny cohort by hand: one row per patient.
cohort = pd.DataFrame({
    "Patient":      ["P1", "P2", "P3", "P4", "P5", "P6", "P7", "P8"],
    "Diabetes":     [True,  True,  False, True,  False, True,  False, False],
    "Hypertension": [True,  False, True,  True,  False, False, True,  False],
    "Obesity":      [False, True,  True,  False, False, True,  True,  False],
})
cohort
```

## 2. Turning a condition into a set

A condition such as "has diabetes" picks out a **set**: the patients for whom that condition is `True`. Building this set means going through the table and **collecting** the patients that match.

The code below does exactly that with a simple loop. It looks at one patient at a time, and if that patient has diabetes it adds their name to the set `A`. (`set()` starts an empty set; `.add(...)` puts one item into it.) We build a second set `B` the same way for hypertension.

```python
# Universe: every patient.
U = set(cohort["Patient"])

# Build set A = patients with diabetes, one patient at a time.
A = set()
for i in range(len(cohort)):
    if cohort.loc[i, "Diabetes"] == True:
        A.add(cohort.loc[i, "Patient"])

# Build set B = patients with hypertension.
B = set()
for i in range(len(cohort)):
    if cohort.loc[i, "Hypertension"] == True:
        B.add(cohort.loc[i, "Patient"])

print("U (everyone)     =", U)
print("A (diabetes)     =", A)
print("B (hypertension) =", B)
```

## 3. Subsets, the empty set, and the universe

Two sets can be related. $A$ is a **subset** of $B$, written $A \subseteq B$, when *every* element of $A$ is also in $B$. For example, the set of diabetic patients is a subset of the universe $U$, because every patient is, of course, one of our eight patients.

The **empty set** $\varnothing$ is the set containing nothing. It is a subset of *every* set, there is no element in it that could fail to be in the other set. In Python the empty set is written `set()`, and `<=` tests "is a subset of".

```python
empty = set()

print("Is A a subset of U?             ", A <= U)      # <= means "is a subset of"
print("Is the empty set a subset of A? ", empty <= A)  # the empty set is a subset of everything
print("Number of elements in empty set:", len(empty))
```

## 4. The four set operations

These four operations build new sets from old ones. In plain language:

- **Union** $A \cup B$: everything in $A$ **or** $B$ (or both).
- **Intersection** $A \cap B$: only what is in **both** $A$ and $B$.
- **Difference** $A \setminus B$: what is in $A$ **but not** in $B$.
- **Complement** $A^c = U \setminus A$: everything in the universe that is **not** in $A$.

Notice these mirror the everyday words *or, and, but-not,* and *not*. Python's set symbols read the same way: `|` is union, `&` is intersection, `-` is difference.

```python
print("A union B       (A | B):", A | B)   # in A OR in B
print("A intersect B   (A & B):", A & B)   # in A AND in B
print("A difference B  (A - B):", A - B)   # in A but NOT in B
print("complement of A (U - A):", U - A)   # everyone NOT in A
```

Reading the results back in words:

- **Union** $A \cup B$: patients with diabetes *or* hypertension (or both).
- **Intersection** $A \cap B$: patients with *both*.
- **Difference** $A \setminus B$: diabetes but *not* hypertension.
- **Complement** $A^c$: everyone *without* diabetes.

## 5. Cardinality (counting a set)

The **cardinality** of a set is simply how many elements it has, written $|A|$. In Python that is `len(...)`. Counting these combinations, how many have diabetes, how many have both, how many have either, is the first step toward the Venn-diagram arithmetic in the next notebook.

```python
print("|A| (diabetes)     =", len(A))
print("|B| (hypertension) =", len(B))
print("|A and B| (both)   =", len(A & B))
print("|A or B| (either)  =", len(A | B))
```
