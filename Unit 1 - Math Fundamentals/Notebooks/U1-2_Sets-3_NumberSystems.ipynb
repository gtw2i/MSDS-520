# Sets of Numbers: From Counting to the Real Line

**This notebook teaches the number systems from scratch.**

Not every set is a set of patients — some of the most important sets are sets of **numbers**. Over history, mathematics kept *enlarging* its idea of "number" to solve new problems. You start with the counting numbers; you add negatives when you need to represent debt or loss; you add fractions when you need to divide and share; and finally you fill in the remaining gaps to get the full number line. Each of these collections is a set, and they are **nested** — each new system contains all the previous ones. Knowing which system a quantity lives in tells you what you are allowed to do with it.

## Learning objectives

- Name the number systems $\mathbb{N}, \mathbb{Z}, \mathbb{Q}, \mathbb{R}$ and give examples of each.
- Explain why they are **nested**: $\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$.
- Distinguish **rational** from **irrational** numbers by their decimal expansions.

## Background

- $\mathbb{N}$ — the **natural numbers** $\{0, 1, 2, 3, \dots\}$: the counting numbers.
- $\mathbb{Z}$ — the **integers** $\{\dots, -2, -1, 0, 1, 2, \dots\}$: the naturals together with their negatives.
- $\mathbb{Q}$ — the **rational numbers**: every number that can be written as a ratio $\tfrac{a}{b}$ of two integers (with $b \ne 0$), such as $\tfrac12$, $-\tfrac34$, or $5 = \tfrac{5}{1}$.
- $\mathbb{R}$ — the **real numbers**: every point on the continuous number line, including numbers like $\sqrt2$ and $\pi$ that are **not** ratios of integers.

$$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}.$$

A **rational** number's decimal expansion either **stops** ($\tfrac14 = 0.25$) or **repeats** forever in a pattern ($\tfrac13 = 0.333\ldots$). An **irrational** number's decimal runs on forever with **no repeating pattern** ($\sqrt2$, $\pi$). Together the rationals and irrationals make up all of $\mathbb{R}$.

## 1. The four number systems

Each system adds something the previous one could not express: the integers add **negatives** (so you can go below zero), the rationals add **fractions** (so you can divide), and the reals add the **irrationals** (so you can measure quantities like the diagonal of a square). Below is a handful of numbers drawn from each.

```python
naturals = [0, 1, 2, 3, 4]            # counting numbers (N)
integers = [-2, -1, 0, 1, 2]         # naturals plus negatives (Z)
rationals = [1/2, -3/4, 5, 0.25]     # ratios of integers (Q)
reals = [math.sqrt(2), math.pi, 1/3] # everything on the number line (R)

print("naturals (N): ", naturals)
print("integers (Z): ", integers)
print("rationals (Q):", rationals)
print("reals (R):    ", reals)
```

## 2. Why they nest

Every counting number is also an integer; every integer is a fraction (write it over $1$); and every fraction is a point on the real line. So each system sits **inside** the next:

$$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}.$$

A natural way to picture "inside" is as nested boxes — a smaller box drawn entirely within a larger one.

```python
# --- graphics: nested boxes for N subset Z subset Q subset R (mechanics only) ---
labels = ["R (reals)", "Q (rationals)", "Z (integers)", "N (naturals)"]
sizes = [4, 3, 2, 1]
colors = ["tab:red", "tab:orange", "tab:green", "tab:cyan"]

fig, ax = plt.subplots(figsize=(6, 6))
for label, size, color in zip(labels, sizes, colors):
    ax.add_patch(plt.Rectangle((-size, -size), 2 * size, 2 * size,
                               fill=False, edgecolor=color, lw=2))
    ax.text(0, size - 0.3, label, ha="center", color=color)
ax.set_xlim(-4.5, 4.5)
ax.set_ylim(-4.5, 4.5)
ax.set_aspect("equal")
ax.axis("off")
plt.show()
```

## 3. Rational vs irrational decimals

Here is the practical difference. Write a number as a decimal: if it **stops or eventually repeats** a pattern, it is rational; if it runs on forever **without ever repeating**, it is irrational. Compare $\tfrac13$ (repeats as $0.3333\ldots$) with $\sqrt2$ and $\pi$ (never repeat). The printout below shows the first 15 digits — enough to see $1/3$ settle into a pattern while the other two do not.

```python
print("1/3     =", format(1/3, ".15f"), " (repeats: rational)")
print("sqrt(2) =", format(math.sqrt(2), ".15f"), " (never repeats: irrational)")
print("pi      =", format(math.pi, ".15f"), " (never repeats: irrational)")
```

## 4. Simple membership checks

We can test which system a number belongs to with basic comparisons. A number is a **whole number** if it equals its own rounded value (nothing after the decimal point); it is a **natural number** if it is whole *and* not negative.

```python
def describe(x):
    is_integer = (x == round(x))          # is it a whole number?
    is_natural = is_integer and x >= 0     # whole and non-negative?
    print("value:", x, "| natural?", is_natural, "| integer?", is_integer)

describe(4)
describe(-2)
describe(2.5)
```

## 5. Summary

- The number systems nest: $\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$, each extending the last.
- **Rationals** are ratios $a/b$; their decimals stop or repeat. **Irrationals** (e.g. $\sqrt2, \pi$) never repeat.
- Together, rationals and irrationals fill the **real** number line.
